Missing Data Handling Strategy:

1. Age:
   Median imputation is used because Age is numeric and may contain
   extreme values. Median is less affected by outliers than mean.

2. Fare:
   Median imputation is used because monetary data can be skewed
   by expensive tickets.

3. Sex:
   Mode imputation is used because Sex is categorical.

4. Outcome:
   Rows with missing Outcome are deleted because Outcome is an
   important analysis/target variable and should not be guessed.

5. Travel_Date:
   Forward fill is used after converting values to datetime because
   the records belong to the same event period.

```python
df.loc[~df["Age"].between(0, 100), "Age"] = np.nan

print("Invalid age values converted to missing.")
```

```python
df["Age"].isnull().sum()
```

```python
df["Sex"] = df["Sex"].astype("string").str.strip().str.lower()
```

```python
df["Sex"] = df["Sex"].replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female"
})
```

```python
df["Sex"].value_counts(dropna=False)
```

```python
sex_mode = df["Sex"].mode()[0]

df["Sex"] = df["Sex"].fillna(sex_mode)

print("Missing Sex values filled using mode:", sex_mode)
```

```python
df["Outcome"] = (
    df["Outcome"]
    .astype("string")
    .str.strip()
    .str.lower()
)

df["Outcome"] = df["Outcome"].replace({
    "survived": "Survived",
    "died": "Died"
})

print(df["Outcome"].value_counts())
```

```python
df["Travel_Date"] = pd.to_datetime(
    df["Travel_Date"],
    errors="coerce"
)

print(df["Travel_Date"].dtype)
```

```python
age_median = df["Age"].median()

df["Age"] = df["Age"].fillna(age_median)

print("Age missing values filled with median:", age_median)
```

```python
fare_median = df["Fare"].median()

df["Fare"] = df["Fare"].fillna(fare_median)

print("Fare missing values filled with median:", fare_median)
```

```python
before = len(df)

df = df.dropna(subset=["Outcome"])

after = len(df)

print("Rows removed:", before - after)
```

```python
df["Travel_Date"] = df["Travel_Date"].ffill()

print("Travel_Date forward fill completed.")
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicates before removal:", duplicate_count)
```

```python
df = df.drop_duplicates().copy()
```

```python
def detect_outliers_iqr(data, column):

    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    print("Column:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))

    return outliers
```

```python
age_outliers = detect_outliers_iqr(df, "Age")
display(age_outliers)
```

```python
fare_outliers = detect_outliers_iqr(df, "Fare")
display(fare_outliers)
```

Outlier Handling Decision:

The IQR method was used to identify statistical outliers.

Age:
Impossible ages such as 999 were treated as data-quality errors
and converted to missing values before median imputation.

Fare:
Detected high fare values were retained because they can represent
legitimate expensive tickets. Therefore, they were not deleted or
capped.

This avoids removing valid real-world observations simply because
they are statistically unusual.

```python
df["Passenger_ID"] = df["Passenger_ID"].astype("string")

df["Sex"] = pd.Categorical(
    df["Sex"],
    categories=["Male", "Female"]
)

df["Age"] = df["Age"].astype("float64")

df["Pclass"] = df["Pclass"].astype("Int64")

df["Fare"] = df["Fare"].astype("float64")

df["Travel_Date"] = pd.to_datetime(df["Travel_Date"])

df["Outcome"] = pd.Categorical(
    df["Outcome"],
    categories=["Died", "Survived"]
)

print("Data types corrected successfully.")
```
